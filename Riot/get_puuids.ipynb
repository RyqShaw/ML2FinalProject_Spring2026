```python
import os
from dotenv import load_dotenv

load_dotenv()
KEY = os.environ.get("RIOT_API_KEY") 

if KEY is None:
    print("Error: API_KEY environment variable not set.")
```

```python

REGION = "na1"         
QUEUE = "RANKED_SOLO_5x5"
SAMPLES_PER_TIER = 10     
RATE_LIMIT_DELAY = 1.3 

HEADERS = {"X-Riot-Token": KEY}

TIERS = {
    "IRON":        ["I", "II", "III", "IV"],
    "BRONZE":      ["I", "II", "III", "IV"],
    "SILVER":      ["I", "II", "III", "IV"],
    "GOLD":        ["I", "II", "III", "IV"],
    "PLATINUM":    ["I", "II", "III", "IV"],
    "EMERALD":     ["I", "II", "III", "IV"],
    "DIAMOND":     ["I", "II", "III", "IV"],
}

DIVISIONS = ["I", "II", "III", "IV"]
```

```python
import random

def sample_tier(tier: str, n: int) -> list[dict]:

    all_entries = []

    for division in :
        entries = get_entries_for_division(tier, division, page=1)
        all_entries.extend(entries)
        if len(all_entries) >= n * 3:
            break

    sampled = random.sample(all_entries, min(n, len(all_entries)))
    results = []

    for entry in sampled:
        puuid = entry['puuid']
        if puuid:
            results.append(puuid)

    return results
```

```python
all_results = {}

for tier, division in TIERS, DIVISIONS:
    print(f"[{tier}]")
    results = sample_tier(tier, division SAMPLES_PER_TIER)
    all_results[tier] = results
```
