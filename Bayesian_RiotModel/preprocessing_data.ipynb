```python
import pandas as pd
import datetime
```

## Read in Data

```python
match_data = pd.read_csv("../Riot/match_data.csv", header=None)
match_data.head()
```

## Turn Dictionary entries into individual Columns

```python
DROP_COLS = {
    'championId', 'teamId', 'summoner1Id', 'summoner2Id',
    'item0', 'item1', 'item2', 'item3', 'item4', 'item5', 'item6',
}

processed_match_data = pd.DataFrame()

for index, row in match_data.iterrows():
    player_index = 0
    for item_index in range(len(row)):
        # ignore the first row index for preprocessing
        if item_index == 0 or type(row[item_index]) is float:
            continue
        try:
            # Turn each dict into a data frame row
            game_match = eval(row[item_index], {"datetime": datetime})
            game_match = {k: (int(v) if isinstance(v, bool) else str(v) if k == 'game_start' else v)
                          for k, v in game_match.items() if k not in DROP_COLS}
            game_match['player_id'] = row[0]
            new_match = pd.DataFrame([game_match])
            
            # Concatinate it with previous data
            processed_match_data = pd.concat([processed_match_data, new_match], ignore_index=True)
            player_index += 1
        except Exception as e:
            print(f"Error at row {index}, col {item_index}: {e}")

processed_match_data.head()
```

```python
processed_match_data.dtypes
```

## Aggregate Data

```python
AVG_COLS = [
    'kills', 'deaths', 'assists', 'totalDamageDealtToChampions', 'totalDamageTaken',
    'goldEarned', 'goldSpent', 'totalMinionsKilled', 'neutralMinionsKilled',
    'visionScore', 'wardsPlaced', 'wardsKilled', 'detectorWardsPlaced',
    'dragonKills', 'baronKills', 'turretKills', 'inhibitorKills',
    'totalHeal', 'totalHealsOnTeammates', 'damageSelfMitigated',
    'totalTimeCCDealt', 'timeCCingOthers', 'timePlayed',
]

# take averages of specified columns, for binary classifiers, make them into ratios
player_stats = processed_match_data.groupby('player_id').agg(
    **{col: (col, 'mean') for col in AVG_COLS},
    first_blood_ratio=('firstBloodKill', lambda x: x.sum() / len(x)),
    first_tower_ratio=('firstTowerKill', lambda x: x.sum() / len(x)),
    win_ratio=('win', lambda x: x.sum() / len(x)),
    early_surrender_ratio=('gameEndedInEarlySurrender', lambda x: x.sum() / len(x))
).reset_index()

player_stats.head()
```

## Export for Use in R Bayesian Model

```python
player_stats.to_csv("processed_players.csv", index=False)
```
